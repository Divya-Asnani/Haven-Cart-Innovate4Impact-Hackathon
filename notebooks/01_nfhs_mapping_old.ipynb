```python
import pandas as pd

df_nfhs = pd.read_csv("nfhs_survey.csv")

print(df_nfhs.shape)
df_nfhs.head()
```

```python
print(df_nfhs.columns.tolist())
```

```python
df_nfhs.info()
```

```python
keywords = [
    "violence",
    "physical",
    "sexual",
    "emotional",
    "husband",
    "partner",
    "injury",
    "control",
    "help",
    "alcohol",
    "spousal"
]

mask = df_nfhs["Indicators"].astype(str).str.contains(
    "|".join(keywords),
    case=False,
    na=False
) | df_nfhs[" sub indicators"].astype(str).str.contains(
    "|".join(keywords),
    case=False,
    na=False
)

nfhs_relevant = df_nfhs[mask]

print(nfhs_relevant.shape)
nfhs_relevant[
    ["Indicators", " sub indicators", "NFHS-5 (2019-20)", "STATE/UT"]
].to_string(index=False)
```

```python
# Keep only relevant violence/risk indicators
risk_indicators = df_nfhs[
    df_nfhs["Indicators"].astype(str).str.contains(
        "Gender Based Violence|Tobacco Use and Alcohol",
        case=False,
        na=False
    )
].copy()

risk_indicators[
    ["Indicators", " sub indicators", "NFHS-5 (2019-20)", "STATE/UT"]
].head(30)
```
